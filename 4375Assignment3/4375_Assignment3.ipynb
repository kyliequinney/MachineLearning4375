```python
#upload dataset
import pandas as pd
import numpy as np
df = pd.read_csv('https://github.com/kyliequinney/StudentMathData/raw/refs/heads/main/NBChealth.txt', sep = '|', encoding='latin-1', header=None)
df.head()
```

```python
#drop tweet id and timestamp columns
df = df.drop(columns=[0,1])
df.head()
```

```python
#remove any words that start with '@' symbol
import re

def remove_at_words(text):
  return re.sub(r'@\w+', '', text)

df[2] = df[2].apply(remove_at_words)

df.head(10)
example_with_at = df[df[2].str.contains('@')]
example_with_at
```

```python
#remove '#' symbol from all entries
def remove_hash(text):
  return re.sub(r'#', '', text)

df[2] = df[2].apply(remove_hash)

example_with_hash = df[df[2].str.contains('#')]
example_with_hash #should be emtpy
```

```python
#remove urls
def remove_url(text):
  url_pattern = r"http[s]?://(?:[a-zA-Z]|[0-9]|[$-_@.&+]|[!*\(\),]|(?:%[0-9a-fA-F][0-9a-fA-F]))+"
  return re.sub(url_pattern, '', text)

df[2] = df[2].apply(remove_url)
df.loc[2766, 2] = df.loc[2766, 2].replace('http:', '', 1)
example_with_url = df[df[2].str.contains('http')]
example_with_url #should be empty
```

```python
#convert all to lowercase
df[2] = df[2].str.lower()
df.head(10)
```

```python
#add columns called "centroid" and "error" to df
df['centroid'] = 0
df['error'] = 0
df.head()
#rename column '2' to 'tweet'
df = df.rename(columns={2: 'tweet'})
df.head()
```

```python
def initialize_centroids(data, k):
  centroids = data.sample(n=k, random_state = 0)
  return centroids

#distance = 1 - (intersect of A, B)/ (union of A, B)
def calculate_distance(a, b):
  a_words = a.split()
  b_words = b.split()
  a_int_b = len(set(a_words).intersection(b_words))
  a_union_b = len(set(a_words).union(b_words))

  return 1 - (a_int_b / a_union_b)

#assign observation to closest centroid
def assign_clusters(data, centroids):
  k = centroids.shape[0]
  n = data.shape[0]
  clusters = []
  assign_errors = []

  for i in range(n):
    all_errors = np.array([])
    for centroid in range(k):
      error = calculate_distance(centroids.iloc[centroid]['tweet'], df.iloc[i]['tweet'])
      all_errors = np.append(all_errors, error)
    nearest_cluster = np.where(all_errors==np.amin(all_errors))[0].tolist()[0]
    nearest_cluster_error = np.amin(all_errors)
    clusters.append(nearest_cluster)
    assign_errors.append(nearest_cluster_error)
  return clusters, assign_errors
```

```python
centroids = initialize_centroids(df, 5)
centroids.iloc[0:,0]
```

```python
#k-means algo

#randomly assign centroid for each
centroids = initialize_centroids(df, 235)
initial_centroid_tweets = centroids['tweet'].copy()
loop = True
tot_err = []
while loop:

  #find new location of the centroid by taking mean of all the observations in each cluster
  df['centroid'], df['error'] = assign_clusters(df, centroids)
  tot_err.append(sum(df['error']))

  centroids = df.groupby('centroid').agg({'tweet': 'first','error': 'mean'})
  centroids = centroids.reset_index(drop = True)

  if len(tot_err) > 1:
    if tot_err[len(tot_err)-1] - tot_err[len(tot_err)-2] <= 1e-4:
      loop = False

centroids

```

```python
#SSE
#sum squarred error for each centroid
error_per_cluster = df.groupby('centroid')['error'].sum()
print(error_per_cluster)
#sum error of error_per_cluster
print("SSE: " + str(error_per_cluster.sum()))
```
